```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

theta = np.linspace(0, 12 * np.pi, 1200)
r = np.linspace(0.1, 10, len(theta))

x = r * np.cos(theta)
y = r * np.sin(theta)

points = np.column_stack([x, y])
segments = np.stack([points[:-1], points[1:]], axis=1)


fig, ax = plt.subplots(figsize=(7, 7))
fig.patch.set_facecolor("black")
ax.set_facecolor("black")
ax.set_xlim(-11, 11)
ax.set_ylim(-11, 11)
ax.set_aspect("equal")
ax.axis("off")

spiral = LineCollection([], cmap="hsv", linewidth=2.5)
ax.add_collection(spiral)

def update(frame):
    n = max(2, int(len(segments) * (frame + 1) / 100))
    spiral.set_segments(segments[:n])
    spiral.set_array(np.linspace(0, 1, n))
    return (spiral,)

animation = FuncAnimation(
    fig, update, frames=100, interval=30, blit=False
)

display(HTML(animation.to_jshtml()))
plt.close(fig)
```
