```python
''' A | B   union          everything in either
A & B   intersection   in both
A - B   difference     in A but not in B
A ^ B   symmetric diff in exactly one of them'''

python_users = {"Asha", "Ben", "Chitra", "Dev"}
sql_users    = {"Ben", "Dev", "Esha"}

print("both       :", python_users & sql_users)
print("either     :", python_users | sql_users)
print("only Python:", python_users - sql_users)
print("one skill  :", python_users ^ sql_users)
print("subset?    :", {"Ben"} <= python_users)
```

```python
import time
numbers_list = list(range(200_000))
numbers_set = set(numbers_list)
target = 199_999

start = time.perf_counter()
for _ in range(200):
    target in numbers_list
list_time = time.perf_counter() - start

start = time.perf_counter()
for _ in range(200):
    target in numbers_set
set_time = time.perf_counter() - start

print(f"list: {list_time:.4f} s   set: {set_time:.6f} s")
print(f"the set was about {list_time / max(set_time, 1e-9):,.0f} times faster")
```
