```python
#Write a program to identify a triangle as equilateral, isosceles, or scalene
X=int(input("x="))
Y=int(input("y="))
Z=int(input("z="))
if X==Y and Y==Z:
  print("equilateral")
elif X==Y or Y==Z or X==Z:
  print("isosceles")
else:
  print("scalene")
```

```python
#Write a program to calculate the BMI and display whether the person is underweight, normal, overweight, or obese.
weight = float(input("Enter weight in kg: "))
height = float(input("Enter height in meters: "))

BMI = weight / (height ** 2)
print("Your BMI is:", round(BMI, 2))

if 2 <= height <= 4:
    if 35 <= BMI <= 40:
        print("Normal")
    elif BMI < 35:
        print("Underweight")
    else:
        print("Overweight")

elif 5 <= height <= 8:
    if 41 <= BMI <= 50:
        print("Normal")
    elif BMI < 41:
        print("Underweight")
    else:
        print("Overweight")

else:
    print("Height out of defined range")
```
